```python
import json
from PIL import Image, ImageDraw
import os

# Load the coco.json file
with open('datasets/pigeons_roboflow/training/_annotations.coco.json') as f:
    data = json.load(f)

# Get the list of images from the coco.json file
images = data['images']

# Loop through each image in the dataset
for image in images:
    # Get the image file path
    image_file_path = os.path.join('./datasets/pigeons_roboflow/training', image['file_name'])

    # Open and display the image
    img = Image.open(image_file_path)

    # Get the annotations for this image
    annotations = [ann for ann in data['annotations'] if ann['image_id'] == image['id']]

    # Draw the bounding boxes on the image
    draw = ImageDraw.Draw(img)
    for ann in annotations:
        bbox = ann['bbox']
        draw.rectangle([(bbox[0], bbox[1]), (bbox[0] + bbox[2], bbox[1] + bbox[3])], outline="red")

    # Display the image
    img.show()

    # Print the annotations
    for ann in annotations:
        print(f"Annotations for image {image['file_name']}: {ann}")

    # Wait for user input before moving to the next image
    user_input = input("Press Enter to continue to the next image or 'q' to quit...")
    if user_input.lower() == 'q':
        break
```
